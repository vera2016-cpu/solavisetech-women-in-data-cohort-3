Write each function with a clear name. Add a one-line docstring explaining what it does, then test it with the examples given.
Task 2.1  Write total_value(price, quantity) that returns the price multiplied by the quantity. Test it with the Notebook (price 1500, quantity 40).	[2 marks]

```python
# Task 2.1 functions
def total_value(price, quantity):
    """Calculate the total value of a product."""
    return price * quantity
```

```python
result = total_value(1500, 40)
print(result)
```

```python
# apply_discount
def apply_discount(price, percent=10):
    """Calculate the price after applying a discount."""
    discount = price * percent / 100
    return price - discount
    
# Test using the default 10% discount
print(apply_discount(1500))

# Test using the 25% discount
print(apply_discount(1500, 25))
```

```python
# Task 2.3
def stock_status(quantity):
    """Checking the stock quantity"""
    if quantity == 0:
        return "Out of stock"
    elif quantity < 10:
        return "Low stock"
    else:
        return "In stock"
```

```python
print(stock_status(0))
print(stock_status(5))
print(stock_status(40))
```

```python
# Task 2.4
def products_in_category(inventory, category):
    """Return product names belonging to the specified category."""
    products = []

    for item in inventory:
        if item["category"] == category:
            products.append(item["product"])

    return products


# Test the function
print(products_in_category(inventory, "Electronics"))
```

## quantity contains whole numbers, but pandas shows it as float64. Why do you think this happened
Pandas displays the quantity column as float64 because it may contain missing values (NaN). 
Since standard integer data types cannot represent NaN, Pandas converts the column to float64

```python
def inspect_data(df):
    """Inspect a DataFrame for its size, missing values and duplicates."""
    
    # Number of rows and columns
    print("Number of rows:", df.shape[0])
    print("Number of columns:", df.shape[1])

    # Missing values in each column
    print("\nMissing values in each column:")
    print(df.isnull().sum())

    # Number of duplicate rows
    print("\nNumber of duplicate rows:", df.duplicated().sum())
```
